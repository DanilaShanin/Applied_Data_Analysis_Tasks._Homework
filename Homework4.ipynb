```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import LatentDirichletAllocation, TruncatedSVD, NMF
```

```python
# Загрузка данных
df = pd.read_csv('/content/cookbook.csv')
```

```python
# Извлечение текста из столбца 'text'
texts = df['text'].astype(str).values
```

```python
# Создание векторизатора TF-IDF
vectorizer = TfidfVectorizer(stop_words='english', max_features=10000)
tfidf_matrix = vectorizer.fit_transform(texts)

# LDA
lda_model = LatentDirichletAllocation(n_components=10, random_state=42,max_iter=100)  # 10 тем
lda_topics = lda_model.fit_transform(tfidf_matrix)

# LSA
lsa_model = TruncatedSVD(n_components=10, random_state=42, max_iter=100)  # 10 тем
lsa_topics = lsa_model.fit_transform(tfidf_matrix)

# NMF
# Увеличьте значение параметра max_iter
nmf_model = NMF(n_components=10, random_state=42, max_iter=100)  # 10 тем
nmf_topics = nmf_model.fit_transform(tfidf_matrix)
```

```python
def get_top_words(model, feature_names, n_top_words):
    top_words = []
    for topic_idx, topic in enumerate(model.components_):
        words = [feature_names[i] for i in topic.argsort()[:-n_top_words - 1:-1]]
        top_words.append(words)
    return top_words

feature_names = vectorizer.get_feature_names_out()

```

```python
# Получение топ-слов для каждой темы
top_words_lsa = get_top_words(lsa_model, feature_names, 10)
top_words_nmf = get_top_words(nmf_model, feature_names, 10)
top_words_lda = get_top_words(lda_model, feature_names, 10)

```

```python
# Вывод топ-слов для каждой темы
for idx, topic in enumerate(top_words_lsa):
    print(f"Латентное размещение Дирихле (LDA): тема {idx + 1}:")
    print(", ".join(topic))
    print("\n")

for idx, topic in enumerate(top_words_lsa):
    print(f"Сингулярная декомпозиция (LSA): тема {idx + 1}:")
    print(", ".join(topic))
    print("\n")

for idx, topic in enumerate(top_words_nmf):
    print(f"Негативная факторизация матриц (NMF): тема {idx + 1}:")
    print(", ".join(topic))
    print("\n")
```

```python
import pandas as pd
from natasha import (
    Doc,
    Segmenter,
    MorphVocab,
    NewsEmbedding,
    NewsMorphTagger,
    NewsSyntaxParser,
    NewsNERTagger,
    LOC,
    PER,
    ORG
)
```

```python
# Инициализация инструментов для обработки текста
segmenter = Segmenter()
morph_vocab = MorphVocab()
emb = NewsEmbedding()
morph_tagger = NewsMorphTagger(emb)
syntax_parser = NewsSyntaxParser(emb)
ner_tagger = NewsNERTagger(emb)
names_extractor = NamesExtractor(morph_vocab)
addr_extractor = AddrExtractor(morph_vocab)
```

```python
# Загрузим данные из CSV файла
df = pd.read_csv('/content/cookbook.csv')
```

```python
# Функция для извлечения названий мест, персон и организаций
def extract_entities(text):
    doc = Doc(text)
    doc.segment(segmenter)
    doc.tag_morph(morph_tagger)
    doc.parse_syntax(syntax_parser)
    doc.tag_ner(ner_tagger)

    locations = []
    persons = []
    organizations = []

    for span in doc.spans:
        span.normalize(morph_vocab)

        if span.type == LOC:
            locations.append(span.normal)
        elif span.type == PER:
            persons.append(span.normal)
        elif span.type == ORG:
            organizations.append(span.normal)

    return {
        'locations': list(set(locations)),
        'persons': list(set(persons)),
        'organizations': list(set(organizations))
    }
```

```python
# Приведение значений в столбце 'text' к строке
df['text'] = df['text'].astype(str)
```

```python
# Приложение функции к каждому текстовому полю в DataFrame
entities_df = df.apply(lambda row: extract_entities(row['text']), axis=1, result_type='expand')
```

```python
# Объединение результата с исходным DataFrame
result_df = pd.concat([df, entities_df], axis=1)
# Вывести результат на экран
display(result_df)

# Вывести результат на экран
print(result_df)
# Сохранение результата в новый файл
result_df.to_csv('extracted_entities.csv', index=False)
```
